```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import plotly.graph_objects as go
import plotly.express as px


sns.set()

#-------------------- 차트 관련 속성 (한글처리, 그리드) -----------
plt.rcParams['font.family']= 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python
#df = pd.DataFrame(dict(a=[1,3,2], b=[3,2,1]))
df = sns.load_dataset("penguins")
df.head()
```

```python
df[['bill_length_mm','bill_depth_mm']].plot(figsize=(8,2))
plt.show()
```

```python
df.plot(kind="box",figsize=(8,2))
plt.show()
```

```python
df.plot(kind="bar", figsize=(8,2))
plt.show()
```

```python
df.plot(kind="box", figsize=(8,2))
plt.show()
```

```python
df.plot(kind="hist", figsize=(8,2))
plt.show()
```

```python
df.plot(kind="scatter",x='bill_length_mm',y='bill_depth_mm',figsize=(8,2))
plt.show()
```
