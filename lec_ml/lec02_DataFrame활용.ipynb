* emp [조건절][컬럼]
* emp [컬럼][조건절]

```python
#emp['DEPTNO'] == 30
#emp[emp['DEPTNO'] == 30]
```

```python
emp
```

```python
 emp[emp['DEPTNO'] ==30][['EMPNO', 'ENAME', 'DEPTNO']]
```

```python
emp [(emp['DEPTNO'] == 20) | (emp['DEPTNO']== 30)]
emp [ emp['DEPTNO'].isin([20,30])]
```

```python
emp [~ (emp['DEPTNO'].isin( [20,30]))]
```

* <b>DEPTNO가 10인 사원 출력

```python
emp[emp['DEPTNO'] == 10]
```

* <b>DEPTNO가 10이고 급여가 2000이상 사원  출력

```python
emp[(emp['DEPTNO'] == 10) & (emp['SAL'] >= 2000)]
```

* <b>DEPTNO가 10이고 급여가 2000이상 사원의 EMPNO, ENAME, SAL, DEPTNO 출력

```python
emp[(emp['DEPTNO'] == 10) & (emp['SAL'] >= 2000)] [['EMPNO', 'ENAME','SAL', 'DEPTNO']]
```
