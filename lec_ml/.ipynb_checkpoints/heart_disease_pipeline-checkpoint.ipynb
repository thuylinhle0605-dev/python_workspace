## 2.2. Xử lý ngoại lệ (Outlier)

**Ý tưởng:** Ngoại lệ là giá trị **hợp lệ nhưng nằm quá xa** phần còn lại — ví dụ cholesterol 564 mg/dl trong khi đa số
bệnh nhân ở mức 200–300. Với mô hình tuyến tính như Logistic Regression, vài điểm cực trị có thể **kéo lệch đường phân tách**.

**Cách phát hiện — quy tắc IQR:**
- `Q1` = phân vị 25%, `Q3` = phân vị 75%, `IQR = Q3 - Q1` (khoảng chứa 50% dữ liệu ở giữa).
- Mọi điểm nằm ngoài `[Q1 - 1.5*IQR, Q3 + 1.5*IQR]` được coi là ngoại lệ. Đây chính là hai "râu" của biểu đồ boxplot.

**Cách xử lý — cắt ngưỡng (capping) chứ không xoá dòng:** giá trị vượt ngưỡng được kéo về đúng ngưỡng biên.
Vì sao không xoá? Vì bệnh nhân nặng thường nằm ở đuôi phân phối — xoá đi là mất chính nhóm mà mô hình cần học nhất.
Capping vẫn giữ thông tin "người này có chỉ số rất cao" mà loại bỏ được ảnh hưởng cực đoan.

⚠️ **Lưu ý cực kỳ quan trọng:** ngưỡng cắt phải **tính trên tập train rồi lưu lại** để dùng y hệt cho tập test.
Nếu tính ngưỡng riêng cho tập test là ta đã "nhìn trộm" dữ liệu tương lai — lỗi **rò rỉ dữ liệu (data leakage)**.

````python
NUM_COLS = ["Age", "BP", "Cholesterol", "Max HR", "ST depression"]   # các cột liên tục

def get_bounds(data, cols):
    """Tính ngưỡng trên/dưới theo quy tắc IQR cho từng cột."""
    bounds = {}
    for c in cols:
        q1, q3 = data[c].quantile([0.25, 0.75])
        iqr = q3 - q1
        bounds[c] = (q1 - 1.5 * iqr, q3 + 1.5 * iqr)
    return bounds

def cap_outliers(data, bounds):
    """Kéo các giá trị vượt ngưỡng về đúng ngưỡng biên."""
    out = data.copy()
    for c, (low, high) in bounds.items():
        out[c] = out[c].clip(low, high)
    return out

BOUNDS = get_bounds(train, NUM_COLS)      # <- tính MỘT LẦN trên tập train, dùng lại cho tập test

for c, (low, high) in BOUNDS.items():
    n_outlier = ((train[c] < low) | (train[c] > high)).sum()
    print(f"{c:15s} ngưỡng [{low:7.1f}, {high:7.1f}]  ->  {n_outlier:6,} ngoại lệ")
````

````python
train_clean = cap_outliers(train, BOUNDS)

# So sánh boxplot trước và sau khi cắt ngưỡng
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
train[NUM_COLS].plot(kind="box", ax=axes[0], rot=20)
axes[0].set_title("Trước: còn nhiều điểm ngoài râu")
train_clean[NUM_COLS].plot(kind="box", ax=axes[1], rot=20)
axes[1].set_title("Sau: các điểm cực trị đã bị kéo về ngưỡng")
plt.tight_layout()
plt.show()

print("Số dòng trước và sau:", len(train), "->", len(train_clean), "(capping không làm mất dòng nào)")
````

# Bước 3. Feature Engineering (Kỹ thuật đặc trưng)

Bước này biến bảng dữ liệu thô thành **ma trận số mà mô hình hiểu đúng**, rồi giữ lại những đặc trưng thực sự có ích. Gồm 2 việc:
mã hoá biến phân loại và chọn lọc đặc trưng.

## 3.1. Mã hoá biến phân loại — One-Hot Encoding

**Ý tưởng:** Cột `Chest pain type` nhận giá trị 1, 2, 3, 4. Nhìn thì là số, nhưng đó chỉ là **mã định danh cho 4 kiểu đau ngực**,
không phải thứ tự. Nếu để nguyên, mô hình tuyến tính sẽ ngầm hiểu rằng *"kiểu 4 lớn gấp đôi kiểu 2"* và *"khoảng cách
từ kiểu 1 đến 2 bằng khoảng cách từ 3 đến 4"* — hoàn toàn sai.

**Giải pháp One-Hot:** tách mỗi mức thành một cột nhị phân riêng.

| Chest pain type | → | type_2 | type_3 | type_4 |
|:---:|:---:|:---:|:---:|:---:|
| 1 | | 0 | 0 | 0 |
| 2 | | 1 | 0 | 0 |
| 4 | | 0 | 0 | 1 |

Ta dùng `drop_first=True` để **bỏ bớt cột của mức đầu tiên**: mức 1 đã được biểu diễn ngầm bằng "cả ba cột đều bằng 0".
Giữ đủ 4 cột sẽ tạo ra quan hệ "tổng 4 cột luôn bằng 1" khiến mô hình không xác định được hệ số duy nhất — gọi là **bẫy biến giả**.

**Lưu ý:** `Number of vessels fluro` (0→3 mạch tổn thương) **có** ý nghĩa thứ bậc thật (càng nhiều càng nặng)
nên giữ nguyên dạng số, không one-hot.

````python
def make_features(data):
    """Biến bảng dữ liệu thô thành ma trận đặc trưng số."""
    X = data.drop(columns=[c for c in ["id", TARGET, "target"] if c in data.columns])
    return pd.get_dummies(X, columns=CAT_COLS, drop_first=True, dtype=int)

X = make_features(train_clean)
y = train_clean["target"]

print(f"13 đặc trưng gốc  ->  {X.shape[1]} cột sau One-Hot Encoding")
print()
print(list(X.columns))
X.head()
````

## 3.2. Chọn lọc đặc trưng — giữ lại cái gì thực sự có ích

**Ý tưởng:** Thêm đặc trưng không phải lúc nào cũng tốt. Một đặc trưng không liên quan gì đến nhãn chỉ mang lại **nhiễu**:
nó làm mô hình phức tạp hơn, dễ học nhầm vào ngẫu nhiên của tập train, và khiến kết quả khó giải thích hơn.

**Công cụ: Mutual Information (MI) — lượng tin tương hỗ.** MI trả lời câu hỏi:
> *"Nếu tôi biết giá trị của đặc trưng này, tôi giảm được bao nhiêu phần bất định về nhãn?"*

- `MI = 0` ⇒ biết đặc trưng cũng **không giúp gì** cho việc đoán nhãn.
- `MI` càng lớn ⇒ đặc trưng càng nhiều thông tin.

**Vì sao dùng MI thay vì tương quan?** Tương quan chỉ bắt được quan hệ dạng **đường thẳng**. MI bắt được **mọi dạng quan hệ**,
kể cả phi tuyến — ví dụ "nguy cơ cao ở cả người rất trẻ lẫn người rất già" thì tương quan ≈ 0 nhưng MI vẫn lớn.

**Cách làm:** tính MI cho từng đặc trưng, loại những đặc trưng có MI gần như bằng 0, rồi **kiểm chứng bằng cross-validation**
rằng việc loại bỏ này không làm mô hình kém đi.

````python
# MI tính hơi lâu nên dùng mẫu 50.000 dòng là đủ ổn định
sample_idx = X.sample(50_000, random_state=RANDOM_STATE).index
mi = mutual_info_classif(X.loc[sample_idx], y.loc[sample_idx], random_state=RANDOM_STATE)
mi = pd.Series(mi, index=X.columns).sort_values(ascending=False)

plt.figure(figsize=(9, 6))
plt.barh(mi.index[::-1], mi.values[::-1], color="#2e86c1")
plt.axvline(0.001, color="red", ls="--", label="ngưỡng loại bỏ = 0.001")
plt.xlabel("Mutual Information với nhãn")
plt.title("Đặc trưng nào mang nhiều thông tin về bệnh tim nhất?")
plt.legend()
plt.tight_layout()
plt.show()

print(mi.round(4).to_string())
````

````python
SELECTED = mi[mi >= 0.001].index.tolist()      # giữ lại đặc trưng có MI đủ lớn
DROPPED = mi[mi < 0.001].index.tolist()

print(f"Giữ lại {len(SELECTED)} đặc trưng.")
print("Loại bỏ :", DROPPED)
````

**Đừng vội tin!** Việc loại đặc trưng phải được **kiểm chứng bằng số liệu**, không phải bằng cảm tính.

Ô dưới đây so sánh điểm F1 của hai phương án bằng **cross-validation 3-fold**: chia tập train thành 3 phần, lần lượt
lấy 2 phần để học và 1 phần để chấm điểm, rồi lấy trung bình. Cách này cho ước lượng ổn định hơn nhiều so với chia một lần.

Nếu điểm không giảm đáng kể, ta chọn phương án **ít đặc trưng hơn** — theo nguyên tắc: *trong các mô hình tốt ngang nhau,
hãy chọn mô hình đơn giản nhất*.

````python
def make_model():
    """Mô hình = chuẩn hoá dữ liệu + Logistic Regression (giải thích kỹ ở Bước 4)."""
    return Pipeline([
        ("scaler", StandardScaler()),
        ("logreg", LogisticRegression(max_iter=1000, random_state=RANDOM_STATE)),
    ])

f1_all = cross_val_score(make_model(), X, y, cv=3, scoring="f1", n_jobs=-1).mean()
f1_sel = cross_val_score(make_model(), X[SELECTED], y, cv=3, scoring="f1", n_jobs=-1).mean()

print(f"Dùng toàn bộ {X.shape[1]} đặc trưng : F1 = {f1_all:.4f}")
print(f"Dùng {len(SELECTED)} đặc trưng đã chọn : F1 = {f1_sel:.4f}")
print(f"Chênh lệch: {f1_all - f1_sel:+.4f}")

FEATURES = SELECTED
print(f"\n=> Bỏ bớt {len(DROPPED)} đặc trưng gần như không ảnh hưởng tới hiệu năng,")
print(f"   nên ta dùng tập {len(FEATURES)} đặc trưng cho gọn và dễ giải thích hơn.")
````

# Bước 4. Huấn luyện mô hình

## 4.1. Ý tưởng — 3 câu hỏi nền tảng

**(a) Vì sao phải tách tập train và tập test?**

Chấm điểm mô hình trên chính dữ liệu nó đã học giống như cho học sinh thi lại đúng đề đã ôn — điểm cao nhưng không nói lên
năng lực thật. Ta giữ riêng **20% dữ liệu mà mô hình chưa từng thấy** để chấm điểm trung thực.

Tham số `stratify=y` giữ **tỉ lệ hai lớp ở tập train và test giống hệt nhau**. Không có nó, tập test có thể lệch lớp
và mọi độ đo trở nên méo mó.

**(b) Vì sao phải chuẩn hoá dữ liệu (StandardScaler)?**

Các đặc trưng có đơn vị rất khác nhau: `Cholesterol` cỡ hàng trăm, `ST depression` chỉ từ 0 đến 6. Điều này gây 2 vấn đề:
- Logistic Regression học bằng **gradient descent**; thang đo lệch nhau khiến quá trình hội tụ chậm và chệch choạc.
- Nó dùng **regularization** (phạt hệ số lớn để chống học vẹt); đặc trưng có thang đo lớn sẽ bị phạt oan.

`StandardScaler` đưa mọi cột về cùng thang: `z = (x - trung bình) / độ lệch chuẩn` ⇒ trung bình 0, độ lệch chuẩn 1.

**(c) Vì sao gói vào Pipeline?**

`Pipeline` xâu chuỗi các bước thành **một mô hình duy nhất**. Điều này không chỉ cho gọn mà còn **chống rò rỉ dữ liệu**:
trung bình và độ lệch chuẩn dùng để chuẩn hoá chỉ được tính từ tập train. Nếu tính trên toàn bộ dữ liệu, thông tin từ
tập test đã "rò rỉ" sang quá trình huấn luyện và điểm số thu được sẽ đẹp hơn thực tế.

```
Pipeline:   dữ liệu thô  ->  [StandardScaler]  ->  [LogisticRegression]  ->  dự đoán
```

## 4.2. Code — Tách tập train/test

````python
X_train, X_test, y_train, y_test = train_test_split(
    X[FEATURES], y,
    test_size=0.2,              # 80% để học, 20% để chấm điểm
    stratify=y,                 # giữ nguyên tỉ lệ 2 lớp ở cả hai tập
    random_state=RANDOM_STATE,
)

print(f"Tập train: {len(X_train):,} dòng")
print(f"Tập test : {len(X_test):,} dòng")
print()
print(f"Tỉ lệ lớp dương ở train: {y_train.mean():.4f}")
print(f"Tỉ lệ lớp dương ở test : {y_test.mean():.4f}   <- stratify giúp hai con số này bằng nhau")
````

## 4.3. Code — Huấn luyện mô hình

Chỉ cần **2 dòng**: tạo pipeline rồi gọi `.fit()`. Bên trong, scikit-learn tự động làm tuần tự:

1. `StandardScaler` học trung bình và độ lệch chuẩn **từ tập train**, rồi chuẩn hoá `X_train`.
2. `LogisticRegression` học các hệ số trên dữ liệu đã chuẩn hoá.

Sau khi huấn luyện, mô hình cho ra **xác suất mắc bệnh** qua `predict_proba()`. Hàm `predict()` chỉ đơn giản so xác suất đó
với **ngưỡng 0.5**: lớn hơn thì kết luận có bệnh. Ô code dưới đây in cả hai để bạn thấy rõ mối liên hệ này.

````python
model = make_model()
model.fit(X_train, y_train)

print("Đã huấn luyện xong. Cấu trúc mô hình:")
print(model)

# Thử với 5 bệnh nhân đầu tiên trong tập test
print("\nXác suất mắc bệnh:", model.predict_proba(X_test.head())[:, 1].round(3))
print("Nhãn dự đoán     :", model.predict(X_test.head()))
print("Nhãn thực tế     :", y_test.head().values)
````

# Bước 5. Đánh giá mô hình

## 5.1. Ý tưởng — hiểu đúng 4 độ đo

Mọi độ đo đều sinh ra từ **ma trận nhầm lẫn (Confusion Matrix)** — bảng đếm 4 tình huống có thể xảy ra:

| | Mô hình đoán: Không bệnh | Mô hình đoán: Có bệnh |
|---|---|---|
| **Thực tế: Không bệnh** | **TN** — đoán đúng | **FP** — báo động giả |
| **Thực tế: Có bệnh** | **FN** — bỏ sót bệnh nhân | **TP** — đoán đúng |

Từ 4 con số đó:

$$\text{Accuracy}=\frac{TP+TN}{\text{tổng số ca}} \qquad
\text{Precision}=\frac{TP}{TP+FP} \qquad
\text{Recall}=\frac{TP}{TP+FN} \qquad
F_1=2\cdot\frac{P \cdot R}{P+R}$$

**Diễn giải bằng lời:**

- **Accuracy** — *"Mô hình đoán đúng bao nhiêu phần trăm số ca?"* Con số tổng quát, dễ hiểu, nhưng vô dụng khi dữ liệu mất cân bằng.
- **Precision** — *"Trong những người bị mô hình chẩn đoán có bệnh, bao nhiêu người thực sự có bệnh?"*
  Precision thấp ⇒ nhiều **báo động giả** ⇒ tốn kém xét nghiệm, gây hoang mang.
- **Recall** — *"Trong những người thực sự có bệnh, mô hình phát hiện được bao nhiêu?"*
  Recall thấp ⇒ nhiều bệnh nhân **bị bỏ sót** ⇒ trong y tế đây là lỗi nguy hiểm nhất.
- **F1-score** — trung bình điều hoà của Precision và Recall. Nó chỉ cao khi **cả hai** cùng cao, nên dùng làm chỉ số tổng hợp.

**Precision và Recall luôn đánh đổi nhau:** muốn bắt hết người bệnh (Recall cao) thì phải "nghi ngờ" nhiều hơn,
kéo theo nhiều báo động giả (Precision giảm). Vì vậy đừng bao giờ chỉ báo cáo một trong hai — luôn đi kèm nhau,
và F1 là cách gộp chúng lại thành một con số duy nhất.

## 5.2. Code — Tính 4 độ đo trên tập test

````python
y_pred = model.predict(X_test)      # dự đoán trên tập test - dữ liệu mô hình CHƯA TỪNG THẤY

print("KẾT QUẢ TRÊN TẬP TEST")
print("-" * 40)
print(f"Accuracy  : {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision : {precision_score(y_test, y_pred):.4f}")
print(f"Recall    : {recall_score(y_test, y_pred):.4f}")
print(f"F1-score  : {f1_score(y_test, y_pred):.4f}")
````

## 5.3. Đọc kết quả như thế nào?

Đừng dừng ở việc ghi lại 4 con số — hãy dịch chúng thành câu nói về bệnh nhân:

- **Accuracy ≈ 0.89** → cứ 100 bệnh nhân thì mô hình xếp đúng khoảng 89 người.
- **Precision ≈ 0.88** → trong 100 người **bị mô hình chẩn đoán có bệnh**, khoảng 88 người thực sự có bệnh,
  12 người còn lại là **báo động giả** (phải làm thêm xét nghiệm không cần thiết).
- **Recall ≈ 0.86** → trong 100 người **thực sự có bệnh**, mô hình phát hiện được khoảng 86 người,
  **bỏ sót 14 người**. Đây là con số đáng lo nhất trong một bài toán y tế.
- **F1 ≈ 0.87** → điểm tổng hợp, cho thấy mô hình cân bằng tương đối tốt giữa hai loại sai sót.

**Câu hỏi để suy nghĩ:** với bài toán sàng lọc bệnh tim, bạn sẵn sàng chấp nhận thêm bao nhiêu ca báo động giả
để giảm số ca bỏ sót? Câu trả lời là một **quyết định nghiệp vụ của bác sĩ**, không phải quyết định kỹ thuật của
người làm mô hình — nhưng người làm mô hình phải đưa ra được con số để bác sĩ cân nhắc.

# Bước 6. Dự đoán trên dữ liệu mới

**Ý tưởng:** Mô hình chỉ có giá trị khi dùng được trên dữ liệu thật. Tập `test.csv` chưa có nhãn — đây chính là tình huống
khi triển khai thực tế: bệnh nhân mới đến khám, ta cần dự đoán ngay.

**Nguyên tắc vàng:** dữ liệu mới phải đi qua **đúng chuỗi biến đổi mà tập train đã đi qua**, với **đúng những tham số đã học từ train**:

```
dữ liệu mới -> cắt ngưỡng bằng BOUNDS (của train) -> One-Hot -> chọn đúng FEATURES -> model.predict()
```

Tuyệt đối không tính lại ngưỡng IQR hay trung bình/độ lệch chuẩn trên dữ liệu mới. Đó là **rò rỉ dữ liệu** —
lỗi phổ biến nhất khiến mô hình chạy tốt trong notebook nhưng thất bại khi triển khai.

````python
test_clean = cap_outliers(test, BOUNDS)                 # dùng lại BOUNDS đã học từ train
X_new = make_features(test_clean).reindex(columns=FEATURES, fill_value=0)

pred_new = model.predict(X_new)
result = pd.DataFrame({
    "id": test["id"],
    TARGET: np.where(pred_new == 1, "Presence", "Absence"),
})
result.to_csv("submission.csv", index=False)

print(f"Đã dự đoán {len(result):,} bệnh nhân và lưu vào 'submission.csv'")
print()
print("Tỉ lệ nhãn dự đoán (nên gần với tỉ lệ ở tập train ~45% Presence):")
print((result[TARGET].value_counts(normalize=True) * 100).round(2))
result.head()
````

## Toàn bộ pipeline trong một hình

```
train.csv
   |
   |-- Bước 1: xác định bài toán  ->  phân loại nhị phân, lớp dương = "Presence"
   |
   |-- Bước 2: kiểm tra chất lượng -> cắt ngưỡng ngoại lệ (IQR) -> EDA bằng 4 biểu đồ
   |
   |-- Bước 3: One-Hot Encoding  ->  chọn lọc đặc trưng bằng Mutual Information
   |
   |-- Bước 4: chia train/test (stratify)  ->  Pipeline[StandardScaler + LogisticRegression]  ->  fit
   |
   |-- Bước 5: đánh giá bằng Accuracy / Precision / Recall / F1
   |
   '-- Bước 6: áp dụng đúng chuỗi biến đổi đó lên test.csv  ->  submission.csv
```

## 5 điều cần nhớ

1. **Xác định đúng loại bài toán trước khi code.** Đầu ra rời rạc 2 lớp ⇒ binary classification ⇒ dùng các độ đo phân loại.
2. **Không bao giờ để dữ liệu test ảnh hưởng tới quá trình huấn luyện.** Mọi tham số biến đổi (ngưỡng IQR, trung bình,
   độ lệch chuẩn) đều chỉ được học từ tập train. Dùng `Pipeline` là cách an toàn nhất để bảo đảm điều này.
3. **Một con số Accuracy không đủ để kết luận.** Luôn báo cáo kèm Precision và Recall, vì hai kiểu sai sót
   (báo động giả và bỏ sót bệnh nhân) có hậu quả hoàn toàn khác nhau.
4. **Precision và Recall luôn đánh đổi nhau.** Chọn nghiêng về bên nào là **quyết định nghiệp vụ**, không phải quyết định kỹ thuật.
5. **Mọi quyết định nên được kiểm chứng bằng số liệu**, ví dụ dùng cross-validation để xác nhận việc bỏ bớt đặc trưng
   không làm mô hình kém đi.
