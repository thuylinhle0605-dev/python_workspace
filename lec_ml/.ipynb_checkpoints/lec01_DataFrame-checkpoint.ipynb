## 결측 삭제
* <font color=red>df.dropna(axis=0) : 행(줄,가로,↓)지우기
* <font color=red>df.dropna(axis=1) : 열(칸,세로,→)지우기
* <font color=red>df.dropna(subset=['컬럼']) : 컬럼에 결측 지우기
* <font color=red>df.drop(columns=['컬럼']) : 컬럼 삭제

# ***★★★Grouping*★★★**
* max(), min(), sum(), count(), std()

```python
df77 = pd.DataFrame(
    {
        "A": ["one", "one", "two", "three"] * 3,
        "B": ["A", "B", "C"] * 4,
        "C": ["foo", "foo", "foo", "bar", "bar", "bar"] * 2,
        "D": np.random.randn(12),
        "E": np.random.randn(12),
    }
)
df77.head(3)
```

```python
df77.groupby("A")["D"].sum()
```

```python
df77.groupby("A")["D"].min()
```

```python
df77.groupby("A")["D"].agg(['min','max'])
```

```python
df77.groupby("A").agg({"D":"min", "E":"max"})
```

# **merge vs. concat**

## concat(axis=)
* <font color=red><B>0 : 위(2) 아래(3)  = 총 5줄
* <font color=red><B>1 : 열(2) 열(3)    = 총5열

```python
dfa = pd.DataFrame({"a" : [1,2,3] , "b":[10,20,30]})
dfb = pd.DataFrame({"a" : [4,5] , "b":[40,50]})
```

```python
dfa
```

```python
dfb
```

```python
pd.concat([dfa, dfb], axis=0)
```

```python
dfab = pd.concat([dfa, dfb], axis=0)
dfab = dfab.reset_index(drop=True)
dfab
```

```python
dfab = pd.concat([dfa, dfb], axis=1)
dfab.columns = ['a','b', 'c','d']
dfab
```

## **merge(on="columns",how=inner/outer)**
* <font color=red><B>0 : 위(2) 아래(3)  = 총 5줄
* <font color=red><B>1 : 열(2) 열(3)    = 총5열


```python
dfab= dfa.merge(dfb, left_index=True, right_index=True, how='inner')
dfab
```

```python
left = pd.DataFrame({"key": ["foo", "foo"], "lval": [1, 2]})
right = pd.DataFrame({"key": ["foo", "foo"], "rval": [4, 5]})
```

```python
left
```

```python
right
```

```python
dfab= left.merge(right,on='key', how='inner')
dfab
```

```python
left = pd.DataFrame({"key": ["foo", "foo2"], "lval": [1, 2]})
right = pd.DataFrame({"key": ["foo", "foo2"], "rval": [4, 5]})
```

```python
dfab1= left.merge(right,on="key", how='inner')
dfab1
```
