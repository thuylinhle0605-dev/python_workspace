# **컨프리헨션 & 람다**

* <b>for문 1~10 짝수만 출력

```python
for v in range(1,11):
    if(v%2==0):
        print(v, end = ' ')
```

```python
[v for v in range(1,11) if (v%2==0) ]       
```

```python
* <b>for문 1~5 3보다크면 값
```

```python
[v*2  if(v>3) else v for v in range(1,6)]
```

```python
for v in range (1,6):
    if v>3:
        print(v*2, end=' ')
    else:
        print(v, end= ' ')
```

# **람다(lambda)**
* 익명(이름없는) 함수
* 1회용

* 함수이름 : add
* 파라미터 : 숫자2개 
* 합산 결과 리턴

* 람다 예제

```python
def add(a,b):
    res= a + b
    return res
m= add(3,5)
print(m)
```

```python
add2= lambda a, b : a+b
add2(3,5)
```

```python
def ncheck(n):
    print(n*2)
ncheck(6)
```

```python
myf = lambda n :n*2
```

```python
myf(6)
```

```python
a = lambda n : n if n > 10 else None
a(20)
```

# **DataFrame을 활용한 lambda**

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({
    "name": ["kim", "hong", "park"],
    "age": [20, 25, 30]
})
df
```

```python
df['age'] = df['age'] + 10
df
```

```python
df['age']= df['age'].apply(lambda a : a+10)
df
```

```python
df['etc']= df['age'].apply(lambda a : '성인'if a>=20 else '미성인' )
df
```
