# **제문어:WHILE**

*1부터 5까지 while문 사용해 출력

````python
num = 1
while(True):
    if (num >5):
        break
    else :
        print(num, end = "\t")
        num = num + 1
````

````python
num = 1
while(num <= 5) :
    print(num, end= "\t ")
    num = num + 1
````
