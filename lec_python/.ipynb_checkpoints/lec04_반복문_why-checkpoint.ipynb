```python
addr_list = [['kim', '010'], ['hong', '999'], ['park', '111']]
```

```python
for n, t in addr_list:
    if n == 'hong':
        addr_list.remove([n,t])

print(addr_list)
```

```python
addr_list=[]
while(True) :
    print("___________________________")    print("1:등록    2:수정   3:삭제   4:조회   Q: 종료")    menu = input("메뉴를 선택하세요")
    
        if  ( menu in ["Q", "q"] )  :        print("종료")        break
    elif ( menu =='1'):
        name = input("이름을 입력하세요")
        tel = input("전화번호 입력하세요")

        addr_list.append(   [name]    )
        print(f"{name}:{tel} 저장되였습니다")
        
    elif ( menu =='2'):
        print('수정')
        search_name = input("수정하려면 이름을 입력하세요")
        update_tel  = input("변경된 전화번호 입력하세요")
        for v in addr_list :
            if v[0] == search_name :
                v[1] == update_tel
    elif (menu =='3'):
        print('삭제')
        search_name = input("삭제하려면 이름을 입력헤주세요")
        for v in addr_list:
            if v[0] == search_name :
                addr_list.remove(v)
                
    elif (menu =='4'):
        print(f" {len(addr_list)}건 조회")
        for v in addr_list:
            print(v)    else :         print(f"잘못된 메뉴 번호를 선택하셨습니디." )
        
```
