리스트에서 20 보다 작은 3의 배수를 출력하라리스트 = [13, 21, 12, 14, 30, 18]

```python
mylist = [13, 21, 12, 14, 30, 18]
for v in mylist:
    if( v < 20 and v % 3==0):
        print(v)
        
```

```python
for dan in range(2,10):
    print(f"{dan}단", end="\t")
    for gob in range(1,10) :
        print(f"{dan}*{gob}={dan*gob}", end="\t")
    print()
```
