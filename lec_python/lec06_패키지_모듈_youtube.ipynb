```python
import requests
import json
```

```python

url = "http://www.gimhae.go.kr/openapi/tour/restaurant.do"

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/151.0.0.0 Safari/537.36"
    ),
    "Accept": "application/json, text/javascript, */*; q=0.01",
    "Accept-Language": "ko-KR,ko;q=0.9,en-US;q=0.8,en;q=0.7",
    "Referer": "https://www.gimhae.go.kr/",
}


res = requests.get(
    url,
    headers=headers,
    timeout=10
)

print(res.status_code) #200
data = res.json()

print(data["results"][0])
```

```python
print(data["results"][0]['name'])print(data["results"][0]['category'])print(data["results"][0]['phone'])print(data["results"][0]['address'])
```

```python
for v in data["results"] :
    print(v['name'] , v['category'], v['phone'],v['address']) 
    print("--"*20)
```
