# Phân tích các thư viện Machine Learning

## 1. Chia dữ liệu

| Thư viện / Hàm | Chức năng | Mục đích |
|---|---|---|
| `train_test_split` | Chia dữ liệu | Chia dữ liệu thành tập Train và Test |

---

## 2. Mô hình Classification

| Model | Loại | Chức năng |
|---|---|---|
| `LogisticRegression` | Linear Model | Phân loại bằng mô hình tuyến tính |
| `RandomForestClassifier` | Tree Model | Phân loại bằng nhiều Decision Tree |
| `VotingClassifier` | Ensemble | Kết hợp nhiều mô hình Classification |

---

## 3. Mô hình Ensemble

| Model | Loại bài toán | Chức năng |
|---|---|---|
| `VotingClassifier` | Classification | Kết hợp kết quả của nhiều model phân loại |
| `VotingRegressor` | Regression | Kết hợp kết quả của nhiều model hồi quy |

---

## 4. Đánh giá Model

| Hàm | Chỉ số | Ý nghĩa |
|---|---|---|
| `accuracy_score` | Accuracy | Tỷ lệ dự đoán đúng trên tổng số dữ liệu |
| `precision_score` | Precision | Trong số dự đoán Positive, bao nhiêu là Positive thật |
| `recall_score` | Recall | Trong số Positive thực tế, model phát hiện đúng bao nhiêu |
| `f1_score` | F1-score | Cân bằng giữa Precision và Recall |
| `precision_recall_curve` | Precision-Recall Curve | Phân tích mối quan hệ giữa Precision và Recall |
| `roc_auc_score` | ROC-AUC | Đánh giá khả năng phân biệt giữa các class |
| `roc_curve` | ROC Curve | Tạo đường cong ROC |
| `confusion_matrix` | Confusion Matrix | Phân tích dự đoán đúng và sai |
| `classification_report` | Classification Report | Hiển thị Precision, Recall, F1-score |

---

## 5. Cross Validation

| Hàm | Chức năng |
|---|---|
| `KFold` | Chia dữ liệu thành K phần để kiểm tra chéo |
| `StratifiedKFold` | Kiểm tra chéo và giữ tỷ lệ giữa các class |
| `cross_val_score` | Đánh giá model bằng Cross Validation |
| `cross_validate` | Cross Validation với nhiều chỉ số đánh giá |
| `GridSearchCV` | Tìm Hyperparameter tốt nhất cho model |

---

## 6. Encoding

| Hàm | Chức năng | Ví dụ |
|---|---|---|
| `LabelEncoder` | Chuyển dữ liệu chữ thành số | Nam → 0, Nữ → 1 |
| `OneHotEncoder` | Chuyển dữ liệu category thành các cột 0/1 | Seoul → 1 0 0 |

---

## 7. Scaling

| Hàm | Chức năng |
|---|---|
| `StandardScaler` | Chuẩn hóa dữ liệu theo Mean và Standard Deviation |
| `MinMaxScaler` | Đưa dữ liệu về một khoảng, thường là 0–1 |
| `RobustScaler` | Chuẩn hóa dữ liệu và ít bị ảnh hưởng bởi Outlier |

---

## 8. Dimension Reduction

| Hàm | Chức năng |
|---|---|
| `PCA` | Giảm số chiều của dữ liệu nhưng cố gắng giữ lại thông tin quan trọng |

---
