```python
Git = công cụ quản lý phiên bản (version control).

Nó giúp bạn:

💾 Lưu lại các phiên bản của code.
↩️ Quay lại phiên bản cũ khi code bị lỗi.
👥 Làm việc nhóm mà không ghi đè code của nhau.
🌐 Kết nối với GitHub để lưu và chia sẻ code.
    
```

```python
Git ≠ GitHub: Git là công cụ quản lý phiên bản; GitHub là nền tảng online để lưu trữ và chia sẻ repository Git.
```

```python
👉 Git = quản lý phiên bản
👉 GitHub = nơi lưu và chia sẻ code online
```
