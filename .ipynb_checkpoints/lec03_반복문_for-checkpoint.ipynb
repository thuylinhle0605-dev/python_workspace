## print ( 갑 or 변수 , end= \n)

```python
for v in range(1,11):
    print(v * 10 , end="t")
```

```python
for v in range(1, 11):
    print( v *10 , end="")
    if(v != 10) :
         print("-", end="")
    else :
         print("", end="")
    
```

```python
if(v != 10) :        print("-", end="")    else :        print("", end="")
```

```python
for v in range(1,7):
    print(v *10 , end="")
    if(v !=6):
        print("-", end="")
    else :
        print("",end="")
```

* range 사용
* 1 - 10 짝수면 짝 , 홀수면 홀 출력
* 줄바꿈없이

```python
for v in range(1,11):
    if(v %2==0 ):
       print(v, "짝", end="")
    else :
        print(v, "홀", end="")
    
```

```python
total = 0
for number in range(1,11):
    total = total + number
```

```python
total = 50
for number in range(1,11):
    total = total + number
print(total)
```

* 1부터 10까지 짝수합

```python
total = 0
for v in range(1,11):
    if(v % 2==0):
        total = total + v
print(total)       
```

리스트에서 20 보다 작은 3의 배수를 출력하라리스트 = [13, 21, 12, 14, 30, 18]

```python
mylist = [13, 21, 12, 14, 30, 18]
for v in mylist:
    if( v < 20 and v % 3==0):
        print(v)
        
```
