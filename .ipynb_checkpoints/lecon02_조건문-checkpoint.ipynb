##***if문법***

```python
score=85
if(score>=90 ):
    print("A")
else:
    print("나머지")
```

```python
*90이상이면 A,80이상이면 B, 70이상이면 C,나머지
```

```python
score=85
if (score >=90 ):
    print("A")
elif (score >=80 ):
    print("B")
elif (score >=70 ):
    print("C")
else:
    print("나머지")
```
